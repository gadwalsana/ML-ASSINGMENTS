```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Plot")
plt.show()                                          #AP24110011039
```

```python
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D"]
marks = [80, 70, 90, 85]

plt.bar(students, marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [5, 8, 6, 10, 12]

plt.scatter(x, y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Scatter Plot")
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45, 55, 60, 65, 70, 70, 75, 80, 85, 90]

plt.hist(marks, bins=5)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")
plt.show()                                               #AP24110011039
```

```python
import matplotlib.pyplot as plt

subjects = ["ML", "Java", "OS", "COA"]
marks = [85, 75, 90, 80]

plt.pie(marks, labels=subjects, autopct="%1.1f%%")
plt.title("Subject Marks Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [10, 20, 30, 40, 50]
y2 = [15, 25, 20, 35,45]



plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")

plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Multiple Line Plot")
plt.legend()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.scatterplot(data=df, x="sepal_length",
                y="petal_length", hue="species")

plt.title("Sepal Length vs Petal Length")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.barplot(data=df, x="species", y="sepal_length")

plt.title("Average Sepal Length by Species")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.histplot(data=df, x="sepal_length", bins=10)

plt.title("Sepal Length Distribution")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.boxplot(data=df, x="species", y="petal_length")

plt.title("Petal Length by Species")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")
numeric_df = df.select_dtypes(include="number")
                                                              #AP24110011039
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
