```python
import pandas as pd
import numpy as np

info = pd.DataFrame({
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"],
    "Age": [27.0, 29, np.nan, 32]
})

print(info)
print(info.count())
print("Null values:")
print(info.isnull())
print(info.isnull().sum())
```

```python
import pandas as pd
import numpy as np

info = pd.DataFrame({
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"],
    "Age": [27.0, 29, np.nan, 32]
})

print(info)
print("Sorting order:")
print(info.sort_index())

print("Descending order:")
print(info.sort_index(ascending=False))
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Array:", a)
print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Maximum:", np.max(a))
print("Minimum:", np.min(a))
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker='o')
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Simple Line Graph")
plt.show()
```

```python
from scipy import stats

data = [10, 20, 20, 30, 40, 20, 50]
                                                     #AP24110011039
print("Mean:", stats.tmean(data))
print("Mode:", stats.mode(data))
print("Standard deviation:", stats.tstd(data))
```

```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5]]
y = [10, 20, 30, 40, 50]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[6]])

print("Prediction for 6:", prediction[0])
print("Model coefficient:", model.coef_[0])
print("Model intercept:", model.intercept_)
```
